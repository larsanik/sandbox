```python
# ==========================================
# ЯЧЕЙКА 1: Установка библиотеки (Безопасная)
# ==========================================
!pip install google-play-scraper pandas tqdm -q

# ==========================================
# ЯЧЕЙКА 2: Основной скрипт парсинга Google Play
# ==========================================
# Импорт библиотек с картинки
import json
import logging
import os
import random
import re
import time
from datetime import datetime

import pandas as pd
from bs4 import BeautifulSoup
from google.colab import files
from google_play_scraper import Sort, reviews  # Специализированный парсер GPlay
from tqdm import tqdm

# Настройка логирования
logging.basicConfig(level=logging.INFO)


def fetch_google_play_reviews(package_name, lang="ru", country="ru", count=500):
    """Собирает отзывы из Google Play и скачивает CSV-файл."""
    print(f"Робот запущен. Пакет: {package_name} [{country.upper()}]")
    print(f"Запрос на сбор {count} последних отзывов...")

    all_reviews = []
    continuation_token = None

    # Вычисляем количество итераций (Google Play отдает порциями до 200 штук)
    chunk_size = 200
    iterations = (count // chunk_size) + (1 if count % chunk_size > 0 else 0)

    for i in tqdm(range(iterations), desc="Загрузка отзывов"):
        try:
            # На первом шаге токен пустой, дальше используем полученный маркер
            if continuation_token is None:
                result, continuation_token = reviews(
                    package_name,
                    lang=lang,
                    country=country,
                    sort=Sort.NEWEST,  # Сортировка: сначала новые
                    count=min(chunk_size, count),
                )
            else:
                result, continuation_token = reviews(
                    package_name, continuation_token=continuation_token
                )

            if not result:
                break

            all_reviews.extend(result)

            # Если собрали уже больше, чем просил пользователь — выходим
            if len(all_reviews) >= count:
                all_reviews = all_reviews[:count]
                break

            # Если Google Play больше не отдает токены — отзывы кончились
            if not continuation_token:
                break

            # Имитируем задержку (из библиотек на картинке)
            time.sleep(random.uniform(0.5, 1.5))

        except Exception as e:
            print(f"\n[!] Ошибка при парсинге на шаге {i+1}: {e}")
            break

    if not all_reviews:
        print(
            f"\n[!] Отзывы не найдены. Проверьте имя пакета ({package_name})."
        )
        return

    # Конвертируем полученный JSON-like список словарей в Pandas DataFrame
    raw_df = pd.DataFrame(all_reviews)

    # Приводим структуру к вашему ТЗ
    df = pd.DataFrame()
    df["Автор"] = raw_df.get("userName", "Аноним")
    df["Оценка"] = raw_df.get("score", 0)
    df["Заголовок"] = raw_df.get(
        "title", ""
    )  # В Google Play заголовков часто нет, будет пустым
    df["Текст"] = raw_df.get("content", "")
    df["Дата"] = pd.to_datetime(raw_df.get("at")).dt.strftime(
        "%Y-%m-%d %H:%M:%S"
    )

    # Сохраняем в CSV
    filename = f"gplay_reviews_{package_name}_{country}.csv"
    df.to_csv(filename, index=False, encoding="utf-8-sig")

    print(f"\n[+] Успешно собрано отзывов: {len(df)}")
    print(f"[+] Файл сохранен: {filename}")

    # Скачивание файла на компьютер
    print("[+] Скачивание файла на ваш ПК...")
    files.download(filename)


# ==========================================
# ПАРАМЕТРЫ ЗАПУСКА
# ==========================================
# ID приложения в Google Play (берется из URL после знака ?id=)
TARGET_PACKAGE = (
    "org.telegram.messenger"  # Пример: Telegram. Для WhatsApp: com.whatsapp
)
TARGET_LANG = "ru"  # Язык отзывов
TARGET_COUNTRY = "ru"  # Страна
MAX_REVIEWS = 500  # Сколько отзывов хотите скачать

# Запуск скрипта
fetch_google_play_reviews(
    package_name=TARGET_PACKAGE,
    lang=TARGET_LANG,
    country=TARGET_COUNTRY,
    count=MAX_REVIEWS,
)
```
